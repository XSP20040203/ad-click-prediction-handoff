```python
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.metrics import roc_auc_score
from pandas.core.frame import DataFrame
import warnings
warnings.filterwarnings('ignore')
```

```python
train_ads = pd.read_csv('./data/train/train.csv')
train_feeds = pd.read_csv('./data/train/train_feeds.csv')
test_ads = pd.read_csv('./data/test/final_test.csv')
test_feeds = pd.read_csv('./data/test/test_feeds.csv')
```

```python
# 只使用一些简单id特征
usecols=['log_id', 'label', 'user_id', 'age', 'sex', 'residence', 'device_name',
            'device_size', 'net_type', 'task_id', 'adv_id', 'creat_type_cd']
cols = [f for f in usecols if f not in ['label']]
X_train = train_ads[cols]
y_train = train_ads['label']
X_test = test_ads[cols]
```

```python
model =  CatBoostClassifier()
model.fit(
    X_train,
    y_train,
)
```

```python
# 保存测试集结果
test_pred = model.predict_proba(X_test)[:, 1]
```

```python
# 生成提交文件
df = {'pctr': test_pred}
df = DataFrame(df)
df = df.reset_index()
df.columns = ['id', 'pctr']
df.to_csv('./submission.csv', index=False)
```
